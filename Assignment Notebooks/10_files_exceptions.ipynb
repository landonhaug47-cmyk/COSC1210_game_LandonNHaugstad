### Reading Data from a Text File

```python
with open('accounts.txt', mode='r') as accounts:
    print(f'{"Account":<10}{"Name":<10}{"Balance":>10}')
    bal = 0
    for record in accounts:
        account, name, balance = record.split()
        print(f'{account:<10}{name:<10}{balance:>10}')
        bal = float(balance) + bal  
    print(bal)
```

### Updating a text file

```python
## Updating the file
accounts = open('accounts.txt', 'r')
temp_file = open('temp_file.txt', 'w')
with accounts, temp_file:
    for record in accounts:
        account, name, balance = record.split()
        if account != '300':
            temp_file.write(record)
        else:
            new_record = ' '.join([account, 'Williams', balance])
            temp_file.write(new_record + '\n')
```

```python
# dividebyzero.py
"""Simple exception handling example."""

while True:
    # attempt to convert and divide values
    try:
        number1 = int(input('Enter numerator: '))
        number2 = int(input('Enter denominator: '))
        result = number1 / number2
    except ValueError:  # tried to convert non-numeric value to int
        print('You must enter two integers\n')
    except ZeroDivisionError:  # denominator was 0
        print('Attempted to divide by zero\n')
    else:  # executes only if no exceptions occur
        print(f'{number1:.3f} / {number2:.3f} = {result:.3f}')
        break  # terminate the loop
```

```python
try:
    print('try suite with no exceptions raised')
except:
    print('this will not execute')
else:
    print('else executes because no exceptions in the try suite')
finally:  
    print('finally always executes')
    
```

```python
try:
    print('try suite that raises an exception')
    #int('a')
    print('this executes when no exception')
except ValueError:
    print('a ValueError occurred')
else:
    print('else will not execute because an exception occurred')
finally:  
    print('finally always executes')
```

```python
import pygame
import math
import random

# === CONFIG ===
WIDTH, HEIGHT = 640, 480
MAP_WIDTH, MAP_HEIGHT = 10, 10
TILE_SIZE = 64
FOV = math.pi / 3
NUM_RAYS = 120
MAX_DEPTH = 800
FPS = 60

# === INIT ===
pygame.init()
screen = pygame.display.set_mode((WIDTH, HEIGHT))
clock = pygame.time.Clock()

# === COLORS ===
BLACK = (0, 0, 0)
WHITE = (220, 220, 220)
RED = (255, 0, 0)
GRAY = (100, 100, 100)
GREEN = (0, 255, 0)

# === MAP ===
game_map = [
    "##########",
    "#........#",
    "#..##....#",
    "#........#",
    "#....##..#",
    "#........#",
    "#..##....#",
    "#........#",
    "#........#",
    "##########"
]

def get_tile(x, y):
    if 0 <= x < MAP_WIDTH and 0 <= y < MAP_HEIGHT:
        return game_map[y][x]
    return '#'

# === PLAYER ===
player_x, player_y = 160, 160
player_angle = 0
player_speed = 2

# === ENEMIES ===
enemies = [{'x': 400, 'y': 300, 'dir': 0, 'cooldown': 0}]

# === BULLETS ===
bullets = []

# === RENDERING ===
def cast_rays():
    rays = []
    start_angle = player_angle - FOV / 2
    for ray in range(NUM_RAYS):
        angle = start_angle + ray * FOV / NUM_RAYS
        sin_a = math.sin(angle)
        cos_a = math.cos(angle)

        for depth in range(1, MAX_DEPTH, 5):
            x = player_x + depth * cos_a
            y = player_y + depth * sin_a
            map_x = int(x // TILE_SIZE)
            map_y = int(y // TILE_SIZE)
            if get_tile(map_x, map_y) == '#':
                rays.append((ray, depth, angle))
                break
    return rays

def draw_scene(rays):
    for ray, depth, angle in rays:
        brightness = 255 / (1 + depth * depth * 0.0001)
        color = (brightness, brightness, brightness)
        height = 40000 / (depth + 0.0001)
        pygame.draw.rect(screen, color, (ray * (WIDTH // NUM_RAYS), HEIGHT // 2 - height // 2, WIDTH // NUM_RAYS, height))

def draw_enemies():
    for enemy in enemies:
        dx = enemy['x'] - player_x
        dy = enemy['y'] - player_y
        angle_to_player = math.atan2(dy, dx)
        angle_diff = angle_to_player - player_angle
        if -FOV/2 < angle_diff < FOV/2:
            dist = math.hypot(dx, dy)
            size = 4000 / (dist + 0.0001)
            x_offset = (angle_diff + FOV/2) / FOV * WIDTH
            pygame.draw.rect(screen, RED, (x_offset - size//2, HEIGHT//2 - size//2, size, size))

def move_enemies():
    for enemy in enemies:
        dx = player_x - enemy['x']
        dy = player_y - enemy['y']
        dist = math.hypot(dx, dy)
        if dist > 100:
            enemy['x'] += dx / dist
            enemy['y'] += dy / dist
        enemy['cooldown'] -= 1
        if dist < 200 and enemy['cooldown'] <= 0:
            angle = math.atan2(dy, dx)
            bullets.append({'x': enemy['x'], 'y': enemy['y'], 'vx': math.cos(angle)*4, 'vy': math.sin(angle)*4, 'color': RED})
            enemy['cooldown'] = 60

def move_bullets():
    global enemies
    new_bullets = []
    for bullet in bullets:
        bullet['x'] += bullet['vx']
        bullet['y'] += bullet['vy']
        hit = False
        for enemy in enemies:
            dx = bullet['x'] - enemy['x']
            dy = bullet['y'] - enemy['y']
            dist = math.hypot(dx, dy)
            if dist < 20 and bullet['color'] == GREEN:  # Player bullet hits enemy
                enemies.remove(enemy)
                hit = True
                break
        if not hit:
            new_bullets.append(bullet)
    bullets[:] = new_bullets

def draw_bullets():
    for bullet in bullets:
        dx = bullet['x'] - player_x
        dy = bullet['y'] - player_y
        dist = math.hypot(dx, dy)
        angle = math.atan2(dy, dx) - player_angle
        if -FOV/2 < angle < FOV/2:
            size = 2000 / (dist + 0.0001)
            x_offset = (angle + FOV/2) / FOV * WIDTH
            pygame.draw.circle(screen, bullet['color'], (int(x_offset), HEIGHT//2), int(size))

# === GAME LOOP ===
running = True
while running:
    screen.fill(BLACK)

    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        player_angle -= 0.04
    if keys[pygame.K_RIGHT]:
        player_angle += 0.04
    if keys[pygame.K_w]:
        dx = math.cos(player_angle) * player_speed
        dy = math.sin(player_angle) * player_speed
        if get_tile(int((player_x + dx) // TILE_SIZE), int(player_y // TILE_SIZE)) == '.':
            player_x += dx
        if get_tile(int(player_x // TILE_SIZE), int((player_y + dy) // TILE_SIZE)) == '.':
            player_y += dy
    if keys[pygame.K_s]:
        dx = -math.cos(player_angle) * player_speed
        dy = -math.sin(player_angle) * player_speed
        if get_tile(int((player_x + dx) // TILE_SIZE), int(player_y // TILE_SIZE)) == '.':
            player_x += dx
        if get_tile(int(player_x // TILE_SIZE), int((player_y + dy) // TILE_SIZE)) == '.':
            player_y += dy
    if keys[pygame.K_SPACE]:
        angle = player_angle
        bullets.append({'x': player_x, 'y': player_y, 'vx': math.cos(angle)*6, 'vy': math.sin(angle)*6, 'color': GREEN})

    rays = cast_rays()
    draw_scene(rays)
    draw_enemies()
    draw_bullets()
    move_bullets()
    move_enemies()

    pygame.display.flip()
    clock.tick(FPS)

pygame.quit()
```
